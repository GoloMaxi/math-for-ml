```python
from pathlib import Path
from statistics import median
from time import perf_counter
import sys

import numpy as np


cwd = Path.cwd()
root = next(p for p in (cwd, *cwd.parents) if (p / "src/mathml/vectors.py").is_file())
sys.path.insert(0, str(root / "src"))

from mathml.vectors import norm, pairwise_distances


def pairwise_loop(a, b):
    distances = np.empty((len(a), len(b)), dtype=float)

    for i, point_a in enumerate(a):
        for j, point_b in enumerate(b):
            distances[i, j] = norm(point_a - point_b, 2)

    return distances


def measure(func, a, b, repeats=3):
    times = []
    for _ in range(repeats):
        start = perf_counter()
        func(a, b)
        times.append(perf_counter() - start)
    return median(times)


rng = np.random.default_rng(42)
A = rng.standard_normal((1000, 100))
B = rng.standard_normal((1000, 100))


pairwise_distances(A[:10], B[:10])
pairwise_loop(A[:10], B[:10])

vectorized_time = measure(pairwise_distances, A, B)
loop_time = measure(pairwise_loop, A, B)

print("Евклидовы расстояния: 1000 × 1000 пар, 100 координат")
print(f"Матричная реализация: {vectorized_time:.4f} с")
print(f"Два Python-цикла:     {loop_time:.4f} с")
print(f"Ускорение:            {loop_time / vectorized_time:.1f}×")
```
